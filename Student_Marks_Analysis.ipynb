```python
marks = [78, 85, 92, 67, 88, 76, 95, 81, 73, 90]
total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)
print("Student Marks:", marks)
print("Total Marks:", total)
print("Average Marks:", round(average, 2))
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
```
